```python
latest_year = df["Year"].max()
latest_df = df[df["Year"] == latest_year].copy()

latest_year
```

```python
top_10 = latest_df.nlargest(
    10, "Life evaluation (3-year average)"
)[["Rank", "Country name", "Life evaluation (3-year average)"]]

top_10
```

```python
bottom_10 = latest_df.nsmallest(
    10, "Life evaluation (3-year average)"
)[["Rank", "Country name", "Life evaluation (3-year average)"]]

bottom_10
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_10.sort_values("Life evaluation (3-year average)"),
    x="Life evaluation (3-year average)",
    y="Country name"
)

plt.title(f"Top 10 Happiest Countries — {latest_year}")
plt.xlabel("Life Evaluation (3-year average)")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

bottom_sorted = bottom_10.sort_values(
    "Life evaluation (3-year average)"
)

sns.barplot(
    data=bottom_sorted,
    x="Country name",
    y="Life evaluation (3-year average)"
)

plt.title(f"Bottom 10 Countries by Happiness — {latest_year}")
plt.xlabel("Country")
plt.ylabel("Life Evaluation (3-year average)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

correlation_data = latest_df[
    [
        "Life evaluation (3-year average)",
        gdp_column,
        "Explained by: Social support",
        "Explained by: Healthy life expectancy"
    ]
].corr()

sns.heatmap(
    correlation_data,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title(f"Correlation Between Happiness and Key Factors — {latest_year}")
plt.tight_layout()
plt.show()
```

```python
# Overall correlation matrix

factor_correlation = latest_df[
    [
        "Life evaluation (3-year average)",
        gdp_column,
        "Explained by: Social support",
        "Explained by: Healthy life expectancy"
    ]
].corr()

factor_correlation
```

```python
# Correlation of happiness with key factors

happiness_correlations = factor_correlation[
    "Life evaluation (3-year average)"
].drop("Life evaluation (3-year average)")

happiness_correlations
```

```python
# Sort factors by their correlation with happiness

happiness_correlations.sort_values(ascending=False)
```

```python
# Display the strongest relationship

strongest_factor = happiness_correlations.idxmax()
strongest_correlation = happiness_correlations.max()

print("Strongest positive relationship with happiness:")
print(strongest_factor)
print("Correlation:", round(strongest_correlation, 2))
```

```python
# Project assumptions and limitations

limitations = [
    "The analysis uses the latest available year in the dataset.",
    "Correlation does not prove causation.",
    "The analysis focuses on GDP, social support, and healthy life expectancy.",
    "Country-level averages may hide differences within countries.",
    "The results depend on the quality and definitions of the source dataset."
]

for i, limitation in enumerate(limitations, 1):
    print(f"{i}. {limitation}")
```
